# 2.2. Источники данных и характеристики

**Тема:** Исследование поведения покупателей в онлайн-магазинах для обнаружения паттернов спроса и предпочтений.

В работе рассматриваются открытые источники данных о покупках, товарах и отзывах.

## 1. UCI Online Retail — реляционный набор данных

**Тип:** реляционный набор данных, CSV/XLSX.  
Источник: https://archive.ics.uci.edu/dataset/352/online+retail

**Категориальные:** товар, покупатель, страна.  
**Числовые:** количество, цена, стоимость покупки.  
**Время:** дата покупки.

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00352/Online%20Retail.xlsx"
df = pd.read_excel(url)

print("Размер набора:", df.shape)
df.head()
```

```python
df["TotalPrice"] = df["Quantity"] * df["UnitPrice"]

print("Количество товаров:", df["Description"].nunique())
print("Количество покупателей:", df["CustomerID"].nunique())
print("Количество стран:", df["Country"].nunique())
print("Средняя цена:", round(df["UnitPrice"].mean(), 2))
print("Среднее количество:", round(df["Quantity"].mean(), 2))
print("Общая стоимость покупок:", round(df["TotalPrice"].sum(), 2))
```

## 2. Amazon Reviews — данные об отзывах

**Тип:** иерархически организованные данные.

**Категориальные:** товар, категория, покупатель.  
**Числовые:** оценка, количество полезных голосов, количество голосов.  
**Время:** дата отзыва.

Данные дополняют транзакционную информацию характеристиками потребительских предпочтений.

## 3. Wildberries API — JSON

**Тип:** источник с API, выдача данных в формате JSON.  
Документация: https://dev.wildberries.ru/

**Категориальные:** категория, бренд, товар.  
**Числовые:** просмотры, добавления в корзину, заказы.
